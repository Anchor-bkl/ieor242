This notebook reads training outputs from `out/` and helps produce tables/plots required by the assignment.

```python
from pathlib import Path\n
import json\n
import pandas as pd\n
import matplotlib.pyplot as plt\n
\n
OUT = Path('out')\n
metrics = pd.read_csv(OUT / 'train_metrics.csv')\n
metrics.head()
```

```python
train = metrics[metrics['split'] == 'train']\n
valid = metrics[metrics['split'].str.contains('valid')]\n
\n
plt.figure(figsize=(8,4))\n
plt.plot(train['step'], train['loss'], label='train loss')\n
plt.plot(valid['step'], valid['loss'], marker='o', linestyle='--', label='valid loss')\n
plt.xlabel('step')\n
plt.ylabel('cross-entropy loss')\n
plt.title('Training and Validation Loss')\n
plt.legend()\n
plt.grid(True, alpha=0.3)\n
plt.show()\n
\n
plt.figure(figsize=(8,4))\n
plt.plot(valid['step'], valid['ppl'], marker='o')\n
plt.xlabel('step')\n
plt.ylabel('perplexity')\n
plt.title('Validation Perplexity')\n
plt.grid(True, alpha=0.3)\n
plt.show()
```

```python
print('Latest validation rows:')\n
display(valid.tail(10))\n
\n
print('Overall throughput stats:')\n
display(metrics[['tokens_per_sec', 'elapsed_sec']].dropna().tail(5))
```
